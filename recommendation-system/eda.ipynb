```python
df_interactions = pd.read_parquet('../scraper/data/final/interactions.parquet')
# Los votos no tienen fecha (NaN). to_datetime(..., unit='s') sobre un float
# con NaN dispara un bug de pandas (overflow en np.round). Se convierten solo
# los valores presentes y se reindexa para conservar los NaT.
ts = df_interactions["timestamp"]
df_interactions["timestamp"] = pd.to_datetime(
    ts.dropna().astype("int64"), unit="s"
).reindex(ts.index)
```
